```python
import sympy as sp
```

```python
gamma = sp.symbols(r"\gamma", real=True)
eps = sp.symbols(r"\epsilon", real=True)
t = sp.symbols(r"t", real=True)
qhat = sp.symbols(r"\hat{\rho}_{j}(x) m_{j}(x) E_{j}(x)", real=True)
wbar = sp.symbols(r"\bar{\rho}_j^n \bar{m}_j^n, \bar{E}_j^n", real=True)
```

```python
s_rho = (1-t)*wbar[0] + t*qhat[0]
s_m = (1-t)*wbar[1] + t*qhat[1]
s_E = (1-t)*wbar[2] + t*qhat[2]
```

```python
equation = sp.Eq((gamma - 1) * ( s_E - sp.Rational(1, 2) * s_m**2 / s_rho ), eps)
equation
```

```python
equation = sp.Eq(equation.lhs * s_rho, eps * s_rho)
equation = sp.simplify(equation)
equation
```

```python
expr = equation.lhs - equation.rhs
num, den = sp.together(expr).as_numer_denom()
num
```

```python
poly = sp.Poly(sp.expand(num), t)
poly
```

```python
A = sp.factor(poly.coeff_monomial(t**2))
A
```

```python
B = sp.factor(poly.coeff_monomial(t))
B
```

```python
C = sp.factor(poly.coeff_monomial(1))
C
```

```python
solution1 = (-B + sp.sqrt(B**2 - 4*A*C)) / (2*A)
solution1
```

```python
solution2 = (-B - sp.sqrt(B**2 - 4*A*C)) / (2*A)
solution2
```
