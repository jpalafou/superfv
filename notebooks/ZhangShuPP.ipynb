```python
import sympy as sp
```

```python
gamma = sp.symbols(r"\gamma", real=True)
eps = sp.symbols(r"\epsilon", real=True)
t = sp.symbols(r"t", real=True)
qhat = sp.symbols(r"\hat{\rho}_{j}^\alpha m_{xj}^\alpha m_{yj}^\alpha m_{zj}^\alpha E_{j}^\alpha", real=True)
wbar = sp.symbols(r"\bar{\rho}_j^n \bar{m}_{xj}^n \bar{m}_{yj}^n \bar{m}_{zj}^n \bar{E}_j^n", real=True)
```

```python
s_rho = (1-t)*wbar[0] + t*qhat[0]
s_mx = (1-t)*wbar[1] + t*qhat[1]
s_my = (1-t)*wbar[2] + t*qhat[2]
s_mz = (1-t)*wbar[3] + t*qhat[3]
s_E = (1-t)*wbar[4] + t*qhat[4]
```

```python
equation = sp.Eq((gamma - 1) * ( s_E - sp.Rational(1, 2) * (s_mx**2 + s_my**2 + s_mz**2) / s_rho ), eps)
equation
```

```python
equation = sp.Eq(equation.lhs * s_rho, eps * s_rho)
equation = sp.simplify(equation)
equation
```

```python
expr = equation.lhs - equation.rhs
num, den = sp.together(expr).as_numer_denom()
num
```

```python
poly = sp.Poly(sp.expand(num), t)
poly
```

```python
A = sp.factor(poly.coeff_monomial(t**2))
A
```

```python
B = sp.factor(poly.coeff_monomial(t))
B
```

```python
C = sp.factor(poly.coeff_monomial(1))
C
```

```python
solution1 = (-B + sp.sqrt(B**2 - 4*A*C)) / (2*A)
solution1
```

```python
solution2 = (-B - sp.sqrt(B**2 - 4*A*C)) / (2*A)
solution2
```
